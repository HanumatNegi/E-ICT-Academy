```python
marks = float(input("Enter student's marks: "))
attendance = int(input("Enter student's attendance record: "))

# Student passes if either marks are 7.5 or above OR attendance is 70% or above
if marks >= 7.5 or attendance >= 70:
    print("Eligible")
else:
    print("Not Eligible")
```

```python

marks = float(input("Enter student's marks: "))
attendance = int(input("Enter student's attendance record: "))

# Student must meet at least one main requirement using OR
if marks >= 7.5 or attendance >= 70:

    # Student must also have at least 5 marks AND 50% attendance
    if marks >= 5 and attendance >= 50:
        print("Congratulations, you passed")
    else:
        print("Failed")
else:
    print("Failed")
```
